## 2. Cosine Similarity Whiteboard Activity

This activity was completed by hand, here is how to do it via Python code.

### Inputs for the problem

````python
documents = [
    "Max likes Sam.",    # D1
    "Sam likes Max.",    # D2
    "Max and Sam.",      # D3
    "Winter likes fall." # D4
]

query = "Sam likes winter"
````

### Step 1: Create vocabulary of all terms

First I will concatenate all strings into one.

````python
allText = " ".join(documents + [query])
print(allText)
````

I don't want to have punctuation. There are several ways to do that. Let's see some:

#### Method 1: Iterate over the string, leave out punctuations

````python
import string
print(string.punctuation)
````

````python
allText_noPunct = "".join([ch for ch in allText if not ch in string.punctuation])
print(allText_noPunct)
````

````python
[ch for ch in allText if not ch in string.punctuation]
````

````python
"-".join([ch for ch in allText if not ch in string.punctuation])
````

### Method 2: Use built-in methods

Pythyn's built-in `str` class has many useful methods for translating a set of characters to another one. Let's see the method in action:

````python
allText.translate(str.maketrans("", "", string.punctuation))
````

What's going on?

````python
print(str.maketrans("", "", string.punctuation))
````

We can see that all ASCII Characters from 0-32 and from 65 to 90 are missing, because those characters will be deleted.

Here is what is happening:

```
str.maketrans(from_str, to_str, delete_chars)
#              ^          ^        ^
#           1st arg    2nd arg  3rd arg
```

* 1st argument (""): The characters you want to replace. (Empty string = replace nothing)

* 2nd argument (""): The characters to replace them with. (Empty string = no replacement mapping)

* 3rd argument (string.punctuation): The characters you want to delete entirely.

### Our vocabulary

Here is a Python one-liner that takes our variable `allText_noPunct` and find the unique words in it.

````python
words = set(allText_noPunct.lower().split())
words
````

````python
voc = sorted(words)
voc
````

<h3>We've already created the vocabularity of all unique terms (saved in 'voc' variable)</h3>

<h3>Step 2: represent each document and the query in vector space based on this vocabulary</h3>

````python
doc_vectors = []

num = 1
for doc in documents + [query]:
    doc_dict = {"document": f"d{num}"}

    doc_words = "".join([ch.lower() for ch in doc if not ch in string.punctuation])
    doc_words = doc_words.split()

    word_counts = Counter(doc_words)
    words_vector = {word: word_counts.get(word, 0) for word in voc}

    doc_dict["words_vector"] = words_vector
    doc_vectors.append(doc_dict)

    num += 1

doc_vectors[4]["document"] = "query"
pprint(doc_vectors)
````

````python
doc_ids = [doc["document"] for doc in doc_vectors]
word_vectors = [doc["words_vector"] for doc in doc_vectors]

vectors_df = pd.DataFrame(word_vectors, index=doc_ids)
vectors_df
````

<h3>Step 3: calculate the cosine similarity between the query and each document</h3>

<p>Define cosine similarity function</p>

````python
def cosineSimilarity(vec1, vec2):
    V1 = np.array(vec1)
    V2 = np.array(vec2)
    cosine = np.dot(V1, V2)/(norm(V1)*norm(V2))
    return cosine
````

<p>Calculate cosine similarity for each document</p>

````python
doc_ids = [doc for doc in vectors_df.index.tolist() if doc != "query"]

vectors_df["cs_score"] = None

doc_values = {}
for doc in doc_ids:
    V1 = vectors_df.loc["query"].tolist()[:-1]
    V2 = vectors_df.loc[doc].tolist()[:-1]
    cosine_similarity = cosineSimilarity(V1, V2)
    vectors_df.loc[doc, "cs_score"] = cosine_similarity

scores_df = vectors_df[["cs_score"]].drop("query")
scores_df
````

<h3>Step 4: rank the results based on cosine similarity</h3>

````python
scores_df.sort_values("cs_score", ascending=False)
````
